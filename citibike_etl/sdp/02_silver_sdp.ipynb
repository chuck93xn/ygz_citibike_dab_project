```python
from pyspark import pipelines as dp
```

```python
from pyspark.sql.functions import unix_timestamp, col, to_date
```

```python
@dp.materialized_view(
    comment="Silver layer: cleaned and enriched raw data"
)
def silver_citibike_data():
    df_bronze = spark.table("bronze_citibike_tripdata")

    df_silver = (
        df_bronze
        .withColumn(
            "trip_duration_mins",
            (unix_timestamp(col("ended_at")) - unix_timestamp(col("started_at"))) / 60
        )
        .withColumn(
            "trip_start_date",
            to_date(col("started_at"))
        )
        .select(
            "ride_id",
            "trip_start_date",
            "started_at",
            "ended_at",
            "start_station_name",
            "end_station_name",
            "trip_duration_mins"
        )
    )

    return df_silver
```
